```python
SELECT TOP 100
    ae.id AS activity_entry_id,
    ah.id AS activity_header_id,
    ah.service_type_id,
    ast.description AS service_type_name,
    asv.service_id,
    srv.description AS service_name,
    ae.activity_type_id,
    atp.description AS activity_type_name,
    CONCAT(
        ast.description, ' ',
        srv.description, ' ',
        atp.description
    ) AS cprod_name,
    CONCAT(
        CAST(ah.service_type_id AS VARCHAR),
        '_',
        CAST(asv.service_id AS VARCHAR),
        '_',
        CAST(ae.activity_type_id AS VARCHAR)
    ) AS cprod_src_id,
    asv.is_primary
FROM silver_wip_activityentry ae
LEFT JOIN silver_wip_activityheader ah
    ON ae.activity_header_id = ah.id
LEFT JOIN silver_wip_activityservice asv
    ON ae.activity_service_id = asv.id
LEFT JOIN silver_wip_servicetype ast
    ON ah.service_type_id = ast.id
LEFT JOIN silver_wip_service srv
    ON asv.service_id = srv.id
LEFT JOIN silver_wip_activitytype atp
    ON ae.activity_type_id = atp.id
WHERE asv.is_primary = true
```
