```python
%%sql
CREATE OR REPLACE TEMP VIEW tmp_sys1_service AS
SELECT DISTINCT
    concat(cast(r.id_organisation_source as string), '-', cast(r.service_offered as string)) AS service_src_id,
    c.configured_list_option AS service_src_name,
    cast(r.id_organisation_source as string) AS service_src_sys_inst_id
FROM silver.silver_sone_srreferralin r
LEFT JOIN silver.silver_sone_srconfiguredlistoption c
    ON r.service_offered = cast(c.id as int)
WHERE r.id_organisation_source IS NOT NULL
  AND c.configured_list_option IS NOT NULL
  AND r.removed_data = 0;
```

```python
%%sql
SELECT * FROM tmp_sys1_service LIMIT 50;
```
