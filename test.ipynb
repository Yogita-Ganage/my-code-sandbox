```python
WITH wip_care_product AS (
    SELECT DISTINCT
        CONCAT(
            ast.description, ' ',
            srv.description, ' ',
            atp.description
        ) AS cprod_name,
        'WIP001' AS cprod_src_sys_inst_id,
        CONCAT(
            CAST(ah.service_type_id AS STRING),
            '_',
            CAST(asv.service_id AS STRING),
            '_',
            CAST(ae.activity_type_id AS STRING)
        ) AS cprod_src_id
    FROM silver_wip_activityentry ae
    LEFT JOIN silver_wip_activityheader ah
        ON ae.activity_header_id = ah.id
    LEFT JOIN silver_wip_activityservice asv
        ON ae.activity_service_id = asv.id
    LEFT JOIN silver_wip_servicetype ast
        ON ah.service_type_id = ast.id
    LEFT JOIN silver_wip_service srv
        ON asv.service_id = srv.id
    LEFT JOIN silver_wip_activitytype atp
        ON ae.activity_type_id = atp.id
    WHERE asv.is_primary = true
      AND ah.id IS NOT NULL
      AND ah.service_type_id IS NOT NULL
      AND asv.service_id IS NOT NULL
      AND ae.activity_type_id IS NOT NULL
)

SELECT
    w.cprod_name,
    w.cprod_src_sys_inst_id,
    w.cprod_src_id
FROM wip_care_product w
LEFT JOIN silver_rdm_care_product r
    ON w.cprod_src_sys_inst_id = r.cprod_src_sys_inst_id
   AND w.cprod_src_id = r.cprod_src_id
WHERE r.cprod_src_id IS NULL
```

```python
-- WIP Care Product derivation
-- Purpose:
-- Derive new Care Product records for WIP and return only the 3 required columns:
-- 1. cprod_name
-- 2. cprod_src_sys_inst_id
-- 3. cprod_src_id

WITH wip_care_product AS (

    SELECT DISTINCT

        -- cprod_name:
        -- Monday definition for WIP says:
        -- Service Type + Service Activity + Activity Type
        CONCAT(
            ast.description, ' ',
            srv.description, ' ',
            atp.description
        ) AS cprod_name,

        -- cprod_src_sys_inst_id:
        -- Temporarily hardcoded as WIP001
        -- This can be made dynamic later once Source System Instance reference table is available in Fabric
        'WIP001' AS cprod_src_sys_inst_id,

        -- cprod_src_id:
        -- Monday definition for WIP says:
        -- Service Type ID + Service Activity ID + Activity Type ID
        CONCAT(
            CAST(ah.service_type_id AS STRING),
            '_',
            CAST(asv.service_id AS STRING),
            '_',
            CAST(ae.activity_type_id AS STRING)
        ) AS cprod_src_id

    FROM silver_wip_activityentry ae

    -- Join to activity header to get service_type_id
    LEFT JOIN silver_wip_activityheader ah
        ON ae.activity_header_id = ah.id

    -- Join to activity service to get service_id
    LEFT JOIN silver_wip_activityservice asv
        ON ae.activity_service_id = asv.id

    -- Join to service type lookup to get service type description
    LEFT JOIN silver_wip_servicetype ast
        ON ah.service_type_id = ast.id

    -- Join to service lookup to get service description
    LEFT JOIN silver_wip_service srv
        ON asv.service_id = srv.id

    -- Join to activity type lookup to get activity type description
    LEFT JOIN silver_wip_activitytype atp
        ON ae.activity_type_id = atp.id

    WHERE asv.is_primary = true
      -- Use only primary service mapping for care product derivation
      AND ah.id IS NOT NULL
      AND ah.service_type_id IS NOT NULL
      AND asv.service_id IS NOT NULL
      AND ae.activity_type_id IS NOT NULL
      -- Exclude incomplete/null combinations
)

SELECT
    w.cprod_name,
    w.cprod_src_sys_inst_id,
    w.cprod_src_id
FROM wip_care_product w

-- Compare with already ingested RDM Care Product records
LEFT JOIN silver_rdm_care_product r
    ON w.cprod_src_sys_inst_id = r.cprod_src_sys_inst_id
   AND w.cprod_src_id = r.cprod_src_id

-- Keep only new records not already present in silver_rdm_care_product
WHERE r.cprod_src_id IS NULL
```

```python
-- Create add table for new WIP Care Product records
-- Target table: silver_rdm_care_product_add
-- Only the 3 required columns are populated:
-- 1. cprod_name
-- 2. cprod_src_sys_inst_id
-- 3. cprod_src_id

CREATE OR REPLACE TABLE silver_rdm_care_product_add AS

WITH wip_care_product AS (

    SELECT DISTINCT

        -- cprod_name:
        -- Derived using Service Type + Service + Activity Type
        CONCAT(
            ast.description, ' ',
            srv.description, ' ',
            atp.description
        ) AS cprod_name,

        -- cprod_src_sys_inst_id:
        -- Temporarily hardcoded for WIP
        'WIP001' AS cprod_src_sys_inst_id,

        -- cprod_src_id:
        -- Derived using Service Type ID + Service ID + Activity Type ID
        CONCAT(
            CAST(ah.service_type_id AS STRING),
            '_',
            CAST(asv.service_id AS STRING),
            '_',
            CAST(ae.activity_type_id AS STRING)
        ) AS cprod_src_id

    FROM silver_wip_activityentry ae

    -- Get service type id
    LEFT JOIN silver_wip_activityheader ah
        ON ae.activity_header_id = ah.id

    -- Get service id and primary service flag
    LEFT JOIN silver_wip_activityservice asv
        ON ae.activity_service_id = asv.id

    -- Get service type description
    LEFT JOIN silver_wip_servicetype ast
        ON ah.service_type_id = ast.id

    -- Get service description
    LEFT JOIN silver_wip_service srv
        ON asv.service_id = srv.id

    -- Get activity type description
    LEFT JOIN silver_wip_activitytype atp
        ON ae.activity_type_id = atp.id

    WHERE asv.is_primary = true
      AND ah.id IS NOT NULL
      AND ah.service_type_id IS NOT NULL
      AND asv.service_id IS NOT NULL
      AND ae.activity_type_id IS NOT NULL
)

SELECT
    w.cprod_name,
    w.cprod_src_sys_inst_id,
    w.cprod_src_id
FROM wip_care_product w

-- Exclude records already present in silver_rdm_care_product
LEFT JOIN silver_rdm_care_product r
    ON w.cprod_src_sys_inst_id = r.cprod_src_sys_inst_id
   AND w.cprod_src_id = r.cprod_src_id

WHERE r.cprod_src_id IS NULL
```
