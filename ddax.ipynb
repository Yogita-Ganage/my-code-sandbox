```python
DEFINE
VAR _Core =
    SUMMARIZE(
        'silver_rdm_service_add',
        'silver_rdm_service_add'[service_src_id],
        'silver_rdm_service_add'[service_src_name],
        'silver_rdm_service_add'[service_src_sys_inst_id]
    )

VAR _PrimaryWindowed =
    TOPN(
        2000,
        _Core,
        'silver_rdm_service_add'[service_src_sys_inst_id], 1,
        'silver_rdm_service_add'[service_src_id], 1
    )

EVALUATE
SELECTCOLUMNS(
    _PrimaryWindowed,
    "service_src_id", [service_src_id],
    "service_src_name", [service_src_name],
    "service_src_sys_inst_id", [service_src_sys_inst_id]
)

ORDER BY
    [service_src_sys_inst_id],
    [service_src_id]
```
